# Research Question

Between 2015 and 2024, did U.S. states with more restrictive housing policies— characterized by higher zoning restrictiveness scores, lower housing permit issuance, and slower housing construction growth — experience higher per capita homelessness rates than states with more permissive housing policies, characterized by lower zoning restrictiveness scores, higher permit issuance, and greater housing supply growth? Additionally, did this relationship strengthen over time as housing costs increased?



## Background and Prior Work

Homelessness in the United States has become one of the most prevalent and pressing issues of the 21st century. In early 2024, homelessness reached an all time high of over 700,000 unhoused people which is the largest number since the Housing and Urban Development (HUD) started tracking this. Many people have attributed homelessness to factors at an individual level such as mental illnesses, drug addiction, or unemployment. However, recently there has been a growing body of research that points to structural housing market conditions being a primary factor in the rise of homelessness. In particular, in the book Homelessness is a Housing Problem, Gregg Colburn and Clayton Page Aldern systematically tested a range of conventional explanations, including drug use, poverty rates, weather, whether certain cities offered more protections, and they found that none of these explained why homelessness rates vary across U.S. cities.<sup><a href="#ref1">1</a></sup> Instead, they found that housing market conditions such as cost and availability of rental units were a far more convincing predictor. This finding structures the homelessness problem as a housing supply problem rather than a social services issue.

One of the biggest connectors of housing markets to homelessness is housing costs or the burden of paying rent. The Pew Charitable Trusts compared homelessness and rent data across U.S. metro areas in 2017 and 2022 and found that homelessness is consistently high in cities where rents were also high. The data showed that rent growth slowed and homeless rates declined in those cities. When examining some of these cities that experienced higher rent costs, they found that Austin, a city that experienced around 15 percent growth in housing supply could not keep up with the 22% in housing demand due to the restrictive measures of zoning laws.<sup><a href="#ref2">2</a></sup> These trends motivate our research question because if rising housing costs worsen the homelessness crisis, and zoning laws restrict housing supply which increases housing costs, then it is logical to assume that restrictive zoning laws affect homeless rates, in times of steep rent growth.

Restrictive zoning laws, which include limits to multifamily housing, minimum lot sizes, and height restrictions, are often believed to be the reason for the strain on housing supply and increasing cost, which contributes to the rise of homelessness in large cities. The growing research field investigating how land use regulation affects housing availability and affordability across American cities. For example, economists Joseph Gyourko and Raven Molloy found that metropolitan areas with stricter land regulations tend to have significantly higher housing prices because supply cannot meet the demand.<sup><a href="#ref3">3</a></sup> Since homelessness is closely correlated with housing affordability, this source provides the foundation for studying whether restrictive zoning laws correlate with higher per-capita homelessness. Correspondingly, research comparing metropolitan regions has shown that homelessness rates are substantially higher in more expensive housing markets. This suggests that housing constraints play a factor in homelessness, and it is not just at the individual level.<sup><a href="#ref4">4</a></sup> These findings lay the foundation for our project in examining zoning restrictiveness and how it correlates with homelessness in different cities.

Recent work has directly linked zoning, housing supply, and homelessness outcomes. Homelessness rates across U.S. metropolitan areas are examined, and conclude that regions with tighter housing markets and more constrained development, which is usually tied to zoning rules, consistently demonstrate higher homelessness rates when compared to cities with more flexible housing supply. These new developments emphasize factors like rent levels, vacancy rates, and construction limits instead of individual ones, suggesting that restrictive zoning increases homelessness because of the lack of new housing. In addition to this research, the California Legislative Analyst's Office found that areas with higher housing costs because of constraints experience higher rates of homelessness, particularly in coastal California cities.<sup><a href="#ref5">3</a></sup> These studies will support our hypothesis that zoning restrictions and homelessness are related and not mutually exclusive.

The prior work we have examined only focuses on either cross-sectional comparisons or data on a single city or state, which leaves us to answer the question of the relationship between homelessness and restrictive zoning laws in other cities. We will also examine a specific time period between 2015 and 2025. During this time, housing costs increased in many U.S. areas, and zoning reforms expanded into some cities but not others. Studies show that supply constraints result in demand, which increases housing prices. This suggests that when housing markets tighten, homelessness rises due to the high cost of shelter. Building on previous research, our project will compare U.S. cities with different zoning restrictiveness levels and evaluate whether homelessness per capita is wider or narrower when compared with housing prices, which are a cause of zoning restrictions. By combining longitudinal data with cross-city comparisons, our analysis will include prior work as context, along with detailed data from 2015 to 2025.

### U.S. Census County Population Estimates (2015–2025)**
- **Dataset Name:** Census Bureau County Population Estimates, Vintage 2019 & Vintage 2025
- **Links:**
  - https://www.census.gov/data/tables/time-series/demo/popest/2020s-total-cities-and-towns.html https://data.census.gov/ 
- **Files:** `co-est2019-alldata.csv` (covers 2010–2019) and `co-est2025-alldata.csv` (covers 2020–2025)
- **Number of observations:** ~3,190 counties per year × 11 years = ~35,090 rows after merging and melting to long format
- **Number of variables:** 16 (identifiers + annual estimates) after wrangling
- **Description:** The two raw Census files are merged on the 5-digit FIPS code (state + county) and combined into a single county-year panel covering 2015–2025, saved to `data/02-processed/census_county_2015_2025.csv`. The relevant variables in the combined dataset are:
  - `STNAME`, `CTYNAME`, `fips` — state name, county name, and unique FIPS identifier
  - `REGION`, `DIVISION` — Census geographic groupings
  - `pop` — annual resident population estimate; primary variable for identifying the top 50 most populated counties
  - `net_mig`, `dom_mig`, `intl_mig` — net, domestic, and international migration; proxies for housing demand pressure
  - `births`, `deaths`, `natural_chg` — natural population change components
  - `pop_chg` — total year-over-year population change
- **Shortcomings:**
  - Data is at the county level — U.S. cities do not always align with county boundaries (e.g., NYC spans five counties)
  - Would need to create another column to merge the different datasets, most likely by states because we are shifting the question to be differences from each state. 
  - The two vintage files use slightly different column naming conventions and the 2015–2019 estimates in the older file were modestly revised after the 2020 Census.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ── 1. Load raw files ─────────────────────────────────────────────────────────
raw_2019 = pd.read_csv('data/00-raw/co-est2019-alldata.csv', encoding='latin-1')
raw_2025 = pd.read_csv('data/00-raw/co-est2025-alldata.csv', encoding='latin-1')

print("2019 file shape:", raw_2019.shape)
print("2025 file shape:", raw_2025.shape)

# ── 2. Filter to county level only (SUMLEV 050; SUMLEV 040 = state totals) ───
counties_2019 = raw_2019[raw_2019['SUMLEV'] == 50].copy()
counties_2025 = raw_2025[raw_2025['SUMLEV'] == 50].copy()

# ── 3. Define columns to keep from each file ──────────────────────────────────
id_cols = ['STATE', 'COUNTY', 'STNAME', 'CTYNAME', 'REGION', 'DIVISION']

years_old = [2015, 2016, 2017, 2018, 2019]
years_new = [2020, 2021, 2022, 2023, 2024, 2025]

def get_year_cols_2019(year):
    return {
        f'POPESTIMATE{year}'     : f'pop_{year}',
        f'BIRTHS{year}'          : f'births_{year}',
        f'DEATHS{year}'          : f'deaths_{year}',
        f'NATURALINC{year}'      : f'natural_chg_{year}',
        f'INTERNATIONALMIG{year}': f'intl_mig_{year}',
        f'DOMESTICMIG{year}'     : f'dom_mig_{year}',
        f'NETMIG{year}'          : f'net_mig_{year}',
        f'NPOPCHG_{year}'        : f'pop_chg_{year}',
    }

def get_year_cols_2025(year):
    return {
        f'POPESTIMATE{year}'     : f'pop_{year}',
        f'BIRTHS{year}'          : f'births_{year}',
        f'DEATHS{year}'          : f'deaths_{year}',
        f'NATURALCHG{year}'      : f'natural_chg_{year}',
        f'INTERNATIONALMIG{year}': f'intl_mig_{year}',
        f'DOMESTICMIG{year}'     : f'dom_mig_{year}',
        f'NETMIG{year}'          : f'net_mig_{year}',
        f'NPOPCHG{year}'         : f'pop_chg_{year}',
    }

rename_2019 = {}
for yr in years_old:
    rename_2019.update(get_year_cols_2019(yr))

rename_2025 = {}
for yr in years_new:
    rename_2025.update(get_year_cols_2025(yr))

df_old = (counties_2019[id_cols + list(rename_2019.keys())]
          .rename(columns=rename_2019))

df_new = (counties_2025[id_cols + list(rename_2025.keys())]
          .rename(columns=rename_2025))

# ── 4. Merge the two periods on FIPS county code ──────────────────────────────
for df in (df_old, df_new):
    df['fips'] = df['STATE'].astype(str).str.zfill(2) + \
                 df['COUNTY'].astype(str).str.zfill(3)

df_wide = pd.merge(df_old, df_new, on=['fips'] + id_cols, how='inner')
print(f"\nMerged wide shape: {df_wide.shape}")

# ── 5. Save interim wide file ─────────────────────────────────────────────────
df_wide.to_csv('data/01-interim/census_county_wide.csv', index=False)

# ── 6. Change to tidy (long) format: one row per county × year ─────────────────
keep_metrics = ['pop', 'births', 'deaths', 'natural_chg',
                'intl_mig', 'dom_mig', 'net_mig', 'pop_chg']

df_long = pd.wide_to_long(
    df_wide,
    stubnames=keep_metrics,
    i=['fips'] + id_cols,
    j='year',
    sep='_',
    suffix=r'\d{4}'
).reset_index()

df_long = df_long[df_long['year'].between(2015, 2025)].copy()
df_long['year'] = df_long['year'].astype(int)

print(f"\nTidy long shape: {df_long.shape}")
print(f"Unique counties : {df_long['fips'].nunique()}")
print(f"Unique years    : {sorted(df_long['year'].unique())}")

# ── 7. Missingness analysis ───────────────────────────────────────────────────
missing = df_long.isna().sum()
pct_missing = (missing / len(df_long) * 100).round(2)
miss_df = pd.DataFrame({'missing_n': missing, 'missing_pct': pct_missing})
print("\nMissingness summary:")
print(miss_df[miss_df['missing_n'] > 0])

# Check for systematic missingness in population by year
if df_long['pop'].isna().any():
    print("\nMissing population rows by year:")
    print(df_long[df_long['pop'].isna()].groupby('year').size())
else:
    print("\nNo missing population values.")

# ── 8. Outlier detection on population and net migration ─────────────────────
def flag_outliers(series, k=3.0):
    q1, q3 = series.quantile(0.25), series.quantile(0.75)
    iqr = q3 - q1
    return (series < q1 - k * iqr) | (series > q3 + k * iqr)

df_long['pop_outlier']    = df_long.groupby('year')['pop'].transform(flag_outliers)
df_long['netmig_outlier'] = df_long.groupby('year')['net_mig'].transform(flag_outliers)

print(f"\nPopulation outlier rows (IQR k=3): {df_long['pop_outlier'].sum()}")
print(df_long[df_long['pop_outlier']][['STNAME', 'CTYNAME', 'year', 'pop']]
      .sort_values('pop', ascending=False).head(10))
# These outliers are expected because these counties contain major cities, hence the higher population. 

print(f"\nNet-migration outlier rows: {df_long['netmig_outlier'].sum()}")

# ── 9. Clean: handle missingness ─────────────────────────────────────────────
# Population is complete for all counties and years (Census always estimates it).
# Secondary metrics (births, deaths, migration) may have minor gaps for 2015
# in the merged file. We drop only rows missing the critical variable (pop).
# We do NOT impute migration/demographic flows — filling with means would
# distort the migration signals needed for the zoning-homelessness analysis.
print(f"\nRows before cleaning : {len(df_long)}")
df_clean = df_long.dropna(subset=['pop']).copy()
print(f"Rows after dropna(pop): {len(df_clean)}")


# ── 10. Summary statistics ────────────────────────────────────────────────────
print("\nAll counties, all years — key metrics:")
print(df_clean[keep_metrics].describe().round(1))

print("\nPopulation by year:")
print(df_clean.groupby('year')['pop'].describe().round(0))

print("\nTop 10 counties by 2025 population:")
print(df_clean[df_clean['year'] == 2025]
      .nlargest(10, 'pop')[['STNAME', 'CTYNAME', 'fips', 'pop']]
      .to_string(index=False))

# ── 11. Write final processed file ───────────────────────────────────────────
df_clean.to_csv('data/02-processed/census_county_2015_2025.csv', index=False)
print(f"Final shape: {df_clean.shape}")
print(f"Columns: {list(df_clean.columns)}")

# ── 12. State-level aggregated dataset ───────────────────────────────────────
# Sum all numeric metrics within each state-year; population and flow counts
# are additive so simple aggregation is correct here.
# Useful if we decide to compare state-level effects rather than county-level
df_state = (df_clean
            .groupby(['STATE', 'STNAME', 'REGION', 'DIVISION', 'year'],
                     as_index=False)
            [keep_metrics]
            .sum())

# Rename for clarity
df_state = df_state.rename(columns={
    'STATE'  : 'state_fips',
    'STNAME' : 'state_name',
})

print(f"\nState-level shape: {df_state.shape}")
print(f"Unique states : {df_state['state_fips'].nunique()}")
print(f"Unique years  : {sorted(df_state['year'].unique())}")

print("\nState-level summary statistics:")
print(df_state[keep_metrics].describe().round(1))

print("\nTop 10 states by 2025 population:")
print(df_state[df_state['year'] == 2025]
      .nlargest(10, 'pop')[['state_name', 'state_fips', 'pop']]
      .to_string(index=False))

df_state.to_csv('data/02-processed/census_state_2015_2025.csv', index=False)
print(f"\nSaved → data/02-processed/census_state_2015_2025.csv")
print(f"Final shape: {df_state.shape}")
print(f"Columns: {list(df_state.columns)}")
df_state.head(50)
```

### Wrangling Process Summary**
The raw Excel data was cleaned and structured into a time-series panel using the following steps:
1.  **Header Correction:** The raw file contains 5 rows of descriptive metadata. The import process uses `header=5` to correctly identify the table columns starting at "State, County, FIPS code...".
2.  **Case Normalization:** Column names were standardized to lowercase and stripped of leading/trailing whitespace (e.g., converting `"Year "` to `"year"`) to ensure reliable coding.
3.  **Data Type Enforcement:** The `year` and `hpi` columns were converted to numeric types. This ensures that missing values represented by periods (`.`) are handled as `NaN` (Not a Number) rather than strings.
4.  **Temporal Filtering:** The dataset was filtered to include only observations between the years **2015 and 2025** to align with the project's scope.
5.  **Outlier Flagging:** A grouping logic by `year` was applied to detect extreme price fluctuations using an Interquartile Range (IQR) multiplier of 3.0, identifying counties with anomalous appreciation compared to their peers.

---

### **Dataset Overview**
* **Number of Observations:** \~34,100 rows (approx. 3,100 counties × 11 years).
* **Number of Variables:** 8
* **Output Path:** `data/02-processed/hpi_county_cleaned.csv`

#### **Key Variables**
| Variable Group | Columns | Description |
| :--- | :--- | :--- |
| **Identifiers** | `state`, `county`, `fips code` | Geographic labels and the 5-digit FIPS code used for merging. |
| **Time** | `year` | The calendar year of the index (2015–2025). |
| **Appreciation** | `annual_change_pct` | Year-over-year percentage change in home prices. |
| **Index (Native)** | `hpi` | The main index value (base 100 at the county's first recording). |
| **Index (Fixed)** | `hpi_1990_base`, `hpi_2000_base` | Standardized index values allowing for comparison against fixed historical baselines. |

### **Shortcomings**
* **Developmental Status:** The FHFA classifies the county-level index as "Experimental." It is more susceptible to volatility in rural areas with low transaction volumes than the State or National indexes.
* **Missing Historical Baselines:** Approximately 48% of the `hpi_1990_base` values are missing because those counties lacked sufficient transaction data in 1990. 
* **Nominal vs. Real Values:** The HPI is provided in nominal terms and does not automatically adjust for inflation, requiring manual adjustment if comparing purchasing power across decades.

```python
import pandas as pd
import numpy as np
import os

# ── 1. Load Raw Data ──────────────────────────────────────────────────────────
# header=6 (index 5) is correct for this specific file
df_raw = pd.read_excel('data/00-raw/hpi_at_county.xlsx', header=5)

# CRITICAL: Standardize column names immediately (lowercase and remove spaces)
df_raw.columns = df_raw.columns.str.strip().str.lower()

# ── 2. Filter & Initial Cleaning ──────────────────────────────────────────────
# Now 'year' (lowercase) will definitely work
df_raw['year'] = pd.to_numeric(df_raw['year'], errors='coerce')
df_filtered = df_raw[(df_raw['year'] >= 2015) & (df_raw['year'] <= 2025)].copy()

# The file actually has 8 columns. We map them specifically to avoid size errors.
# 'state', 'county', 'fips', 'year', 'annual_change_pct', 'hpi', 'hpi_1990_base', 'hpi_2000_base'
df_filtered.columns = [
    'state', 'county', 'fips', 'year', 'annual_change_pct', 
    'hpi', 'hpi_1990_base', 'hpi_2000_base'
]

# ── 3. Missingness Analysis ───────────────────────────────────────────────────
missing = df_filtered.isna().sum()
print("Missingness summary (2015-2025):")
print(missing[missing > 0])

# ── 4. Outlier Detection ──────────────────────────────────────────────────────
def flag_outliers(series, k=3.0):
    # Ensure we drop NaNs for the calculation but keep index alignment
    s = series.dropna()
    if s.empty: return pd.Series(False, index=series.index)
    q1, q3 = s.quantile(0.25), s.quantile(0.75)
    iqr = q3 - q1
    return (series < (q1 - k * iqr)) | (series > (q3 + k * iqr))

df_filtered['hpi_outlier'] = df_filtered.groupby('year')['hpi'].transform(flag_outliers)
print(f"\nExtreme HPI Outliers Detected: {df_filtered['hpi_outlier'].sum()}")

# ── 5. Cleaning & Logic ───────────────────────────────────────────────────────
# Dropping the base columns as per your original plan
df_clean = df_filtered.drop(columns=['hpi_1990_base', 'hpi_2000_base'])
df_clean = df_clean.dropna(subset=['hpi']).copy()

print(f"\nFinal cleaned shape: {df_clean.shape}")

# ── 6. Save Processed Data ────────────────────────────────────────────────────
os.makedirs('data/02-processed', exist_ok=True)
df_clean.to_csv('data/02-processed/hpi_county_cleaned.csv', index=False)

# ── 7. Summary Statistics ────────────────────────────────────────────────────
print("\n--- Summary Statistics (Cleaned Data) ---")

# 1. General stats for the main numerical columns
stats = df_clean[['annual_change_pct', 'hpi']].describe().round(2)
print(stats)

# 2. Average HPI per year to see the trend
yearly_trend = df_clean.groupby('year')['hpi'].mean().round(2)
print("\nAverage HPI by Year:")
print(yearly_trend)

# 3. Count of records per state
print("\nTop 5 States by Number of County Records:")
print(df_clean['state'].value_counts().head(5))

# 4. Outlier Summary
outlier_pct = (df_clean['hpi_outlier'].sum() / len(df_clean) * 100).round(2)
print(f"\nOutlier Prevalence: {outlier_pct}% of total records.") 
```

**Description of the variables most relevant to this project:**

The six variables are a part of the explicit growth controls index (EGCI). This is a subindex of zoning restrictiveness that directly restricts supply of housing units, constructed by annual limits and caps within each city.

Limit on Permits (Approval Volume)

- Single-family permits:**  
A cap on the total number of approved standalone construction authorizations issued per year for single, detached households. If the annual limit is 100, the city will not issue a 101st permit that year, completely pausing new detached home projects.

- Multi-family permits:**  
A cap on the total number of multi-unit project approvals processed per year. This restricts the number of apartment, condominium, or townhouse projects allowed to break ground, regardless of whether a project contains 3 units or 300 units.

Limit on Units (Total Housing Capacity)

- Single-family units:
An annual ceiling on individual detached households. Because a single-family structure generally contains exactly one unit, this number is almost always identical to the single-family permit count.

- Multi-family units:
An annual limit on the total number of individual apartments, condos, or suites within multi-family developments. For example, if a city has a cap of 500 multi-family units, a developer trying to build a single 600-apartment building will be blocked because the single project exceeds the entire city's annual allotment.

Limits on Dwellings (Physical Buildings)

- Multi-family dwellings:
A restriction on the total number of physical multi-unit complexes, structures, or properties allowed to be built in a year. It treats an entire 20-story apartment complex as one singular "dwelling structure”.

- Multi-family dwelling units: 
The most legally precise variation of a multi-family unit cap. It specifically limits individual residential spaces equipped with private sleeping, cooking, and sanitation facilities located inside a shared building framework.

**Descriptions of any shortcomings this dataset has with respect to the project**

The data has other variables and also similar datasets available that go more in depth about height restrictions, parking limits, etc. that provides more nuance to “zoning restrictiveness”, but for the purpose of this project and making it easier to work with and calculate, we focused on these aspects of zoning permissions.

There's many cities to work with here, and a lot of extreme differences where some cities have no restrictions and others have restrictions in all six variables. These extreme differences and outliers must be accounted for.

**Other Dataset Information**

Each variable/column for the type of restriction (permit, unit, dwelling) will have a value of 0 or 1. 0 means there is zero restriction or cap on this specific housing growth control, while 1 means there is. The data itself is not specific about which year it came from or when those restrictions were added which may be a problem in analyzing its effect on our dependent variable.

```python
import pandas as pd
import numpy as np
from pathlib import Path

# ---------------------------------------------------
# 1. LOAD RAW DATA
# ---------------------------------------------------

raw_path = Path("data/00-raw/nzlud_muni.csv")

nzlud = pd.read_csv(raw_path, encoding='latin-1')

# preview
print(nzlud.head())


# ---------------------------------------------------
# 2. CHECK IF DATA IS TIDY
# ---------------------------------------------------
# Tidy data rules:
# - each variable = one column
# - each observation = one row
# - each cell = one value

# get only necessary columns
nzlud = nzlud[['GEOID','statename','place','type','restrict_sf_permit', 'restrict_mf_permit', 'limit_sf_units', 'limit_mf_units', 'limit_mf_dwellings', 'limit_mf_dwelling_units']].copy()

# add total restrictions column
nzlud["total_restrictions"] = nzlud[['restrict_sf_permit', 'restrict_mf_permit', 'limit_sf_units', 'limit_mf_units', 'limit_mf_dwellings', 'limit_mf_dwelling_units']].sum(axis=1)

print("\nColumns:")
print(nzlud.columns)

print("\nData Types:")
print(nzlud.dtypes)


# ---------------------------------------------------
# 3. DEMONSTRATE DATASET SIZE
# ---------------------------------------------------

print("\nDataset Shape:")
print(nzlud.shape)

print(f"Rows: {nzlud.shape[0]}")
print(f"Columns: {nzlud.shape[1]}")


# ---------------------------------------------------
# 4. MISSING DATA ANALYSIS
# ---------------------------------------------------

print("\nMissing Values Per Column:")
missing = nzlud.isnull().sum()

print(missing)

print("\nPercent Missing:")
missing_percent = (nzlud.isnull().sum() / len(nzlud)) * 100

print(missing_percent.sort_values(ascending=False))

# rows with ANY missing values
print("\nRows with Missing Values:")
print(nzlud[nzlud.isnull().any(axis=1)].head())

# investigate patterns in missingness
print("\nMissingness Correlation:")
print(nzlud.isnull().corr())

# interpretation guide
print("""
Interpretation:
- Random missingness:
  missing values scattered randomly

- Systematic missingness:
  missingness correlated with certain variables
""")


# ---------------------------------------------------
# 5. FIND OUTLIERS / SUSPICIOUS ENTRIES
# ---------------------------------------------------

numeric_cols = nzlud.select_dtypes(include=np.number).columns

print("\nChecking Numeric Columns for Outliers...")

for col in numeric_cols:

    Q1 = nzlud[col].quantile(0.25)
    Q3 = nzlud[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = nzlud[(nzlud[col] < lower) | (nzlud[col] > upper)]

    print(f"\n{col}")
    print(f"Outliers Found: {len(outliers)}")

# suspicious duplicate rows
print("\nDuplicate Rows:")
print(nzlud.duplicated().sum())


# ---------------------------------------------------
# 6. CLEAN DATA
# ---------------------------------------------------

# remove duplicates
nzlud_clean = nzlud.drop_duplicates()

# drop rows missing ALL values
nzlud_clean = nzlud_clean.dropna(how='all')

# fill numeric missing values with median
for col in numeric_cols:
    nzlud_clean[col] = nzlud_clean[col].fillna(nzlud_clean[col].median())

# fill categorical missing values with mode
cat_cols = nzlud_clean.select_dtypes(exclude=np.number).columns

for col in cat_cols:
    nzlud_clean[col] = nzlud_clean[col].fillna(nzlud_clean[col].mode()[0])

print("\nRemaining Missing Values:")
print(nzlud_clean.isnull().sum().sum())

print("\nCleaned Dataset Shape:")
print(nzlud_clean.shape)


# ---------------------------------------------------
# 7. SAVE FILES
# ---------------------------------------------------

# final processed dataset
processed_path = Path("data/02-processed/nzlud_clean.csv")
nzlud_clean.to_csv(processed_path, index=False)
```

**Description:** This dataset contains annual HUD Point-in-Time homelessness counts by state. Each sheet in the Excel file represents a different year.
```python
# ── 1. Load Raw HUD PIT Data ─────────────────────────────────────────────────
%pip install pyxlsb

file_path = 'data/00-raw/2007-2024-PIT-Counts-by-State.xlsb'
years = list(range(2015, 2025))

keep_cols = [
    "State",
    "Number of CoCs",
    "Overall Homeless",

    # age groups
    "Overall Homeless - Under 18",
    "Overall Homeless - Age 18 to 24",
    "Overall Homeless - Over 24",
    "Overall Homeless - Age 25 to 34",
    "Overall Homeless - Age 35 to 44",
    "Overall Homeless - Age 45 to 54",
    "Overall Homeless - Age 55 to 64",
    "Overall Homeless - Over 64",

    # gender
    "Overall Homeless - Woman",
    "Overall Homeless - Man",

    "Overall Homeless - Transgender",
    "Overall Homeless - Gender Questioning",
    "Overall Homeless - Non Binary",
    "Overall Homeless - Culturally Specific Identity",
    "Overall Homeless - Different Identity",
    "Overall Homeless - More Than One Gender",

    # race / ethnicity
    "Overall Homeless - Non-Hispanic/Latina/e/o",
    "Overall Homeless - Hispanic/Latina/e/o",
    "Overall Homeless - American Indian, Alaska Native, or Indigenous",
    "Overall Homeless - Asian or Asian American",
    "Overall Homeless - Black, African American, or African",
    "Overall Homeless - Middle Eastern or North African",
    "Overall Homeless - Native Hawaiian or Other Pacific Islander",
    "Overall Homeless - White",
    "Overall Homeless - Multi-Racial",
]

# ── 2. Read Each Yearly Sheet ────────────────────────────────────────────────

hud_dfs = []

for year in years:

    temp_df = pd.read_excel(
        file_path,
        sheet_name=str(year),
        engine="pyxlsb"
    )

    # standardize column formatting
    temp_df.columns = temp_df.columns.str.strip()

    # keep only columns that exist for that year
    available_cols = [
        col for col in keep_cols
        if col in temp_df.columns
    ]

    temp_df = temp_df[available_cols].copy()

    # add year column manually
    temp_df["year"] = year

    hud_dfs.append(temp_df)

# ── 3. Combine All Years Into One Raw Dataset ────────────────────────────────

hud_raw = pd.concat(hud_dfs, ignore_index=True)

# force columns into consistent order
final_order = (
    ["State", "year"] +
    [col for col in keep_cols if col != "State" and col in hud_raw.columns]
)

hud_raw = hud_raw[final_order]

print("Raw HUD dataset loaded successfully.")
print("Shape:", hud_raw.shape)
print("Years included:", sorted(hud_raw["year"].unique()))

# preview columns
print("\nColumns:")
for i, col in enumerate(hud_raw.columns):
    print(i, col)
```

```python
# ── 4. Tidy / standardize column names ──────────────────────────────────────

hud_raw.columns = (
    hud_raw.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_", regex=False)
    .str.replace("/", "_", regex=False)
    .str.replace(",", "", regex=False)
)

print("Number of columns:", len(hud_raw.columns))

print("\nColumn names:")
for i, col in enumerate(hud_raw.columns):
    print(i, col)
```

### Merged Dataset: Homelessness, Housing Policy, and Demographics (2015–2024)

To enable a unified analysis, the five cleaned datasets were merged into a single state-year panel. 
First, a state abbreviation-to-full-name mapping was used to standardize the state identifier across 
all datasets, since each used a different format. The HPI data was aggregated from county level to 
state level by computing the mean and median HPI per state per year. The zoning data was aggregated 
from city level to state level by averaging restriction scores across all cities within each state. 
Since the zoning dataset contains no year variable, it was joined as a static attribute, meaning 
each state receives the same zoning scores across all years. The HUD homelessness data and Census 
population data were already at the state level and were merged directly on state and year. Finally, 
a `homeless_per_100k` variable was computed by dividing overall homelessness counts by state population 
and scaling to per 100,000 residents, allowing for fair comparisons across states of different sizes.

```python
import pandas as pd
import numpy as np

# ── State abbreviation → full name mapping ────────────────────────────────────
state_abbr_to_name = {
    'AL': 'Alabama', 'AK': 'Alaska', 'AZ': 'Arizona', 'AR': 'Arkansas',
    'CA': 'California', 'CO': 'Colorado', 'CT': 'Connecticut', 'DE': 'Delaware',
    'FL': 'Florida', 'GA': 'Georgia', 'HI': 'Hawaii', 'ID': 'Idaho',
    'IL': 'Illinois', 'IN': 'Indiana', 'IA': 'Iowa', 'KS': 'Kansas',
    'KY': 'Kentucky', 'LA': 'Louisiana', 'ME': 'Maine', 'MD': 'Maryland',
    'MA': 'Massachusetts', 'MI': 'Michigan', 'MN': 'Minnesota', 'MS': 'Mississippi',
    'MO': 'Missouri', 'MT': 'Montana', 'NE': 'Nebraska', 'NV': 'Nevada',
    'NH': 'New Hampshire', 'NJ': 'New Jersey', 'NM': 'New Mexico', 'NY': 'New York',
    'NC': 'North Carolina', 'ND': 'North Dakota', 'OH': 'Ohio', 'OK': 'Oklahoma',
    'OR': 'Oregon', 'PA': 'Pennsylvania', 'RI': 'Rhode Island', 'SC': 'South Carolina',
    'SD': 'South Dakota', 'TN': 'Tennessee', 'TX': 'Texas', 'UT': 'Utah',
    'VT': 'Vermont', 'VA': 'Virginia', 'WA': 'Washington', 'WV': 'West Virginia',
    'WI': 'Wisconsin', 'WY': 'Wyoming', 'DC': 'District of Columbia'
}
state_name_to_abbr = {v: k for k, v in state_abbr_to_name.items()}

# ── 1. Census: already state-level, just add abbreviation ────────────────────
census = pd.read_csv('data/02-processed/census_state_2015_2025.csv')
census['state'] = census['state_name'].map(state_name_to_abbr)
census = census[census['state'].notna()]  # drop DC if not needed, or keep it
print("Census shape:", census.shape)

# ── 2. HPI: aggregate from county → state level ──────────────────────────────
hpi = pd.read_csv('data/02-processed/hpi_county_cleaned.csv')
hpi_state = (hpi.groupby(['state', 'year'], as_index=False)
               .agg(
                   avg_hpi=('hpi', 'mean'),
                   median_hpi=('hpi', 'median'),
                   avg_annual_change_pct=('annual_change_pct', 'mean')
               ))
print("HPI state shape:", hpi_state.shape)

# ── 3. Zoning: aggregate from city → state level (no year — static join) ─────
zoning = pd.read_csv('data/02-processed/nzlud_clean.csv')
zoning_state = (zoning.groupby('statename', as_index=False)
                      .agg(
                          num_cities=('place', 'count'),
                          avg_total_restrictions=('total_restrictions', 'mean'),
                          pct_cities_with_any_restriction=('total_restrictions', lambda x: (x > 0).mean()),
                          avg_restrict_sf_permit=('restrict_sf_permit', 'mean'),
                          avg_restrict_mf_permit=('restrict_mf_permit', 'mean'),
                          avg_limit_sf_units=('limit_sf_units', 'mean'),
                          avg_limit_mf_units=('limit_mf_units', 'mean'),
                      )
                      .rename(columns={'statename': 'state'}))
print("Zoning state shape:", zoning_state.shape)

# ── 4. HUD: already state-level, drop non-US-state rows ──────────────────────
hud = pd.read_csv('data/02-processed/hud_clean.csv')
hud = hud[hud['state'].isin(state_abbr_to_name.keys())].copy()  # drops "Total", territories
print("HUD shape:", hud.shape)

# ── 5. Merge everything on state + year ───────────────────────────────────────
# Start with HUD (state + year) as base
merged = hud.merge(census, on=['state', 'year'], how='inner')
print("After HUD + Census merge:", merged.shape)

merged = merged.merge(hpi_state, on=['state', 'year'], how='left')
print("After + HPI merge:", merged.shape)

# Zoning has no year — join on state only (applies same zoning score to all years)
merged = merged.merge(zoning_state, on='state', how='left')
print("After + Zoning merge:", merged.shape)

# ── 6. Add homelessness per capita ───────────────────────────────────────────
merged['homeless_per_100k'] = (merged['overall_homeless'] / merged['pop']) * 100_000

print("\nFinal merged shape:", merged.shape)
print("Years:", sorted(merged['year'].unique()))
print("States:", merged['state'].nunique())
print(merged[['state', 'year', 'overall_homeless', 'homeless_per_100k', 
              'avg_hpi', 'avg_total_restrictions']].head(10))

# Drop columns that aren't useful to the research question 
cols_to_drop = [
    'overall_homeless___culturally_specific_identity',
    'overall_homeless___different_identity',
    'overall_homeless___more_than_one_gender',
    'overall_homeless___middle_eastern_or_north_african',
    'overall_homeless___gender_questioning',
    'overall_homeless___non_binary'
]

merged = merged.drop(columns=cols_to_drop)

# Verify
print(merged.isnull().sum()[merged.isnull().sum() > 0])
# ── 7. Save ───────────────────────────────────────────────────────────────────
merged.to_csv('data/02-processed/merged_final.csv', index=False)
print("\nSaved to data/02-processed/merged_final.csv")

# Merged Dataset Preview 
pd.set_option('display.max_columns', None)
merged.head(50)
```

### Geographic Distribution of Homelessness and Zoning Restrictiveness

To begin exploring the relationship between housing policy and homelessness, two choropleth maps were 
produced using state-level averages across 2015–2024. The first map displays homelessness rates per 
100,000 residents by state, while the second displays average zoning restrictiveness scores. Viewing 
these maps side by side allows for a visual comparison of whether states with stricter housing policies 
tend to cluster in the same regions as states with higher homelessness rates.

```python
import geopandas as gpd
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1.inset_locator import inset_axes
import os

# ── Load merged data ──────────────────────────────────────────────────────────
merged = pd.read_csv('data/02-processed/merged_final.csv')

region_labels = {1: 'Northeast', 2: 'Midwest', 3: 'South', 4: 'West'}

merged_avg = merged.groupby('state', as_index=False).agg(
    homeless_per_100k=('homeless_per_100k', 'mean'),
    avg_total_restrictions=('avg_total_restrictions', 'mean'),
    pop=('pop', 'mean'),
    state_name=('state_name', 'first'),
    REGION=('REGION', 'first')
)
merged_avg['region_name'] = merged_avg['REGION'].map(region_labels)

# ── Load shapefile into separate variable ─────────────────────────────────────
url = "https://raw.githubusercontent.com/python-visualization/folium/master/examples/data/us-states.json"
geo_states = gpd.read_file(url)

# ── Merge geo with data ───────────────────────────────────────────────────────
geo_states = geo_states.merge(
    merged_avg[['state', 'homeless_per_100k', 'avg_total_restrictions']],
    left_on='id', right_on='state', how='left'
)

# ── Split into contiguous, AK, HI ────────────────────────────────────────────
exclude    = ['PR', 'GU', 'VI', 'MP', 'AS']
contiguous = geo_states[~geo_states['id'].isin(exclude + ['AK', 'HI'])].copy()
alaska     = geo_states[geo_states['id'] == 'AK'].copy()
hawaii     = geo_states[geo_states['id'] == 'HI'].copy()

def make_map(col, cmap, label, title, filename):

    vmin = geo_states[col].min()
    vmax = geo_states[col].max()

    fig, ax = plt.subplots(1, 1, figsize=(14, 8))

    # ── Main map: contiguous 48 ───────────────────────────────────────────────
    contiguous.plot(
        column=col, ax=ax, cmap=cmap,
        edgecolor='white', linewidth=0.5,
        vmin=vmin, vmax=vmax,
        legend=True,
        missing_kwds={'color': 'lightgrey'},
        legend_kwds={'label': label, 'shrink': 0.5, 'orientation': 'horizontal'}
    )
    ax.set_title(title, fontsize=14, fontweight='bold', pad=15)
    ax.axis('off')

    # ── Alaska inset ──────────────────────────────────────────────────────────
    ax_ak = inset_axes(ax, width='22%', height='22%', loc='lower left',
                       bbox_to_anchor=(0.0, 0.0, 1, 1),
                       bbox_transform=ax.transAxes)
    alaska.plot(column=col, ax=ax_ak, cmap=cmap,
                edgecolor='white', linewidth=0.3,
                vmin=vmin, vmax=vmax)
    ax_ak.set_title('AK', fontsize=8)
    ax_ak.axis('off')
    for spine in ax_ak.spines.values():
        spine.set_visible(True)
        spine.set_edgecolor('gray')
        spine.set_linewidth(0.8)

    # ── Hawaii inset ──────────────────────────────────────────────────────────
    ax_hi = inset_axes(ax, width='15%', height='10%', loc='lower left',
                       bbox_to_anchor=(0.22, 0.0, 1, 1),
                       bbox_transform=ax.transAxes)
    hawaii.plot(column=col, ax=ax_hi, cmap=cmap,
                edgecolor='white', linewidth=0.3,
                vmin=vmin, vmax=vmax)
    ax_hi.set_title('HI', fontsize=8)
    ax_hi.axis('off')
    for spine in ax_hi.spines.values():
        spine.set_visible(True)
        spine.set_edgecolor('gray')
        spine.set_linewidth(0.8)

    plt.tight_layout()
    os.makedirs('results', exist_ok=True)
    plt.savefig(f'results/{filename}', dpi=150, bbox_inches='tight')
    plt.show()
    print(f"Saved to results/{filename}")

make_map(
    col='homeless_per_100k',
    cmap='Reds',
    label='Homeless per 100k',
    title='Average Homelessness Rate per 100k by State (2015–2024)',
    filename='choropleth_homelessness.png'
)

make_map(
    col='avg_total_restrictions',
    cmap='Blues',
    label='Avg Zoning Restrictions',
    title='Average Zoning Restrictiveness Score by State',
    filename='choropleth_zoning.png'
)
```

### Correlational Analysis: Zoning Restrictiveness vs. Homelessness Rate

To directly examine the relationship between housing policy and homelessness, a scatterplot was 
produced plotting each state's average zoning restrictiveness score against its average homelessness 
rate per 100,000 residents, averaged across 2015–2024. Points are sized by state population and 
colored by Census region to reveal whether geographic patterns drive the relationship. A linear 
regression line is overlaid to assess the direction and strength of the association, with the 
correlation coefficient and p-value reported to evaluate statistical significance. This plot 
serves as an initial test of the project's core hypothesis: that states with more restrictive 
housing policies experience higher per capita homelessness rates.

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

fig, ax = plt.subplots(figsize=(12, 8))

palette = {'Northeast': '#e41a1c', 'Midwest': '#377eb8',
           'South': '#4daf4a', 'West': '#ff7f00'}

for region, group in merged_avg.groupby('region_name'):
    sizes = (group['pop'] / group['pop'].max()) * 800 + 50
    ax.scatter(
        group['avg_total_restrictions'],
        group['homeless_per_100k'],
        s=sizes,
        alpha=0.75,
        label=region,
        color=palette[region],
        edgecolors='white',
        linewidth=0.5
    )
    for _, row in group.iterrows():
        ax.annotate(
            row['state'],
            (row['avg_total_restrictions'], row['homeless_per_100k']),
            fontsize=7, ha='center', va='bottom', alpha=0.8
        )

# ── Regression line ───────────────────────────────────────────────────────────
slope, intercept, r, p, se = stats.linregress(
    merged_avg['avg_total_restrictions'],
    merged_avg['homeless_per_100k']
)
x_line = np.linspace(merged_avg['avg_total_restrictions'].min(),
                     merged_avg['avg_total_restrictions'].max(), 100)
ax.plot(x_line, intercept + slope * x_line,
        color='black', linewidth=2, linestyle='--',
        label=f'Regression (r={r:.2f}, p={p:.3f})')

ax.set_xlabel('Average Zoning Restrictiveness Score', fontsize=12)
ax.set_ylabel('Average Homeless per 100k', fontsize=12)
ax.set_title('Zoning Restrictiveness vs. Homelessness Rate by State (2015–2024)\n'
             '(point size = population)', fontsize=13)
ax.legend(title='Region', bbox_to_anchor=(1.01, 1), loc='upper left')
ax.grid(True, alpha=0.3)
plt.tight_layout()
os.makedirs('results', exist_ok=True)
plt.savefig('results/zoning_vs_homelessness_scatter.png', dpi=150, bbox_inches='tight')
plt.show()

print(f"Regression: slope={slope:.2f}, r={r:.2f}, p-value={p:.4f}")
```

## Time-Series Analysis: Homelessness Over Time by Housing Policy 

To study how homelessness changes over time under different housing policies, we created a line graph showing average homelessness per 100,000 people from 2015 to 2024. States were grouped into three categories based on zoning restrictiveness: high, medium, and low (using terciles of the zoning index).

This helps us see whether differences between restrictive and permissive states stay the same or change over time. If zoning restrictiveness limits housing supply, we expect higher homelessness in high-restriction states and a growing gap compared to low-restriction states as housing pressures increase.

We average at the state-year level to make comparisons fair across time and reduce noise from county-level variation. This grouping makes it easier to see long-term patterns. This plot helps test whether housing policy effects stay constant or become stronger over time.

```python
import pandas as pd
import matplotlib.pyplot as plt

# ---------------------------------------------
# LOAD DATA
# ---------------------------------------------
merged = pd.read_csv('data/02-processed/merged_final.csv')

# ---------------------------------------------
# KEEP NEEDED COLUMNS
# ---------------------------------------------
df = merged[['year',
             'avg_total_restrictions',
             'homeless_per_100k']].copy()

# ---------------------------------------------
# REMOVE MISSING VALUES
# ---------------------------------------------
df = df.dropna()

# ---------------------------------------------
# CREATE RESTRICTION GROUPS
# ---------------------------------------------

# manually split into 3 groups

low = df['avg_total_restrictions'].quantile(0.33)
high = df['avg_total_restrictions'].quantile(0.66)

def group_restrictions(x):

    if x <= low:
        return 'Permissive'

    elif x <= high:
        return 'Moderate'

    else:
        return 'Restrictive'

df['restriction_group'] = df['avg_total_restrictions'].apply(group_restrictions)

# ---------------------------------------------
# CALCULATE MEAN HOMELESSNESS BY YEAR
# ---------------------------------------------
grouped = (
    df.groupby(['year', 'restriction_group'])['homeless_per_100k']
      .mean()
      .reset_index()
)

# ---------------------------------------------
# CREATE GRAPH
# ---------------------------------------------
plt.figure(figsize=(12,6))

for group in ['Permissive', 'Moderate', 'Restrictive']:

    subset = grouped[grouped['restriction_group'] == group]

    plt.plot(
        subset['year'],
        subset['homeless_per_100k'],
        marker='o',
        label=group
    )

# ---------------------------------------------
# LABELS
# ---------------------------------------------
plt.title('Homelessness Rates Over Time by Housing Restrictiveness')

plt.xlabel('Year')

plt.ylabel('Average Homelessness per 100k')

plt.legend()

plt.grid(True)

plt.savefig('results/homelessness_rates_over_time_line.png')
plt.show()
```

This helps test whether higher housing prices are linked to higher homelessness rates.
