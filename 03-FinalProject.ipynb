## Background and Prior Work

Homelessness in the United States has become one of the most prevalent and pressing issues of the 21st century. In early 2024, homelessness reached an all time high of over 700,000 unhoused people which is the largest number since the Housing and Urban Development (HUD) started tracking this. Many people have attributed homelessness to factors at an individual level such as mental illnesses, drug addiction, or unemployment. However, recently there has been a growing body of research that points to structural housing market conditions being a primary factor in the rise of homelessness. In particular, in the book Homelessness is a Housing Problem, Gregg Colburn and Clayton Page Aldern systematically tested a range of conventional explanations, including drug use, poverty rates, weather, whether certain cities offered more protections, and they found that none of these explained why homelessness rates vary across U.S. cities.<sup><a href="#ref1">1</a></sup> Instead, they found that housing market conditions such as cost and availability of rental units were a far more convincing predictor. This finding structures the homelessness problem as a housing supply problem rather than a social services issue.

One of the biggest connectors of housing markets to homelessness is housing costs or the burden of paying rent. The Pew Charitable Trusts compared homelessness and rent data across U.S. metro areas in 2017 and 2022 and found that homelessness is consistently high in cities where rents were also high. The data showed that rent growth slowed and homeless rates declined in those cities. When examining some of these cities that experienced higher rent costs, they found that Austin, a city that experienced around 15 percent growth in housing supply could not keep up with the 22% in housing demand due to the restrictive measures of zoning laws.<sup><a href="#ref2">2</a></sup> These trends motivate our research question because if rising housing costs worsen the homelessness crisis, and zoning laws restrict housing supply which increases housing costs, then it is logical to assume that restrictive zoning laws affect homeless rates, in times of steep rent growth.

Restrictive zoning laws, which include limits to multifamily housing, minimum lot sizes, and height restrictions, are often believed to be the reason for the strain on housing supply and increasing cost, which contributes to the rise of homelessness in large cities. The growing research field investigating how land use regulation affects housing availability and affordability across American cities. For example, economists Joseph Gyourko and Raven Molloy found that metropolitan areas with stricter land regulations tend to have significantly higher housing prices because supply cannot meet the demand.<sup><a href="#ref3">3</a></sup> Since homelessness is closely correlated with housing affordability, this source provides the foundation for studying whether restrictive zoning laws correlate with higher per-capita homelessness. Correspondingly, research comparing metropolitan regions has shown that homelessness rates are substantially higher in more expensive housing markets. This suggests that housing constraints play a factor in homelessness, and it is not just at the individual level.<sup><a href="#ref4">4</a></sup> These findings lay the foundation for our project in examining zoning restrictiveness and how it correlates with homelessness in different cities.

Recent work has directly linked zoning, housing supply, and homelessness outcomes. Homelessness rates across U.S. metropolitan areas are examined, and conclude that regions with tighter housing markets and more constrained development, which is usually tied to zoning rules, consistently demonstrate higher homelessness rates when compared to cities with more flexible housing supply. These new developments emphasize factors like rent levels, vacancy rates, and construction limits instead of individual ones, suggesting that restrictive zoning increases homelessness because of the lack of new housing. In addition to this research, the California Legislative Analyst's Office found that areas with higher housing costs because of constraints experience higher rates of homelessness, particularly in coastal California cities.<sup><a href="#ref5">3</a></sup> These studies will support our hypothesis that zoning restrictions and homelessness are related and not mutually exclusive.

The prior work we have examined only focuses on either cross-sectional comparisons or data on a single city, which leaves us to answer the question of the relationship between homelessness and restrictive zoning laws in other states. We will also examine a specific time period between 2015 and 2024. During this time, housing costs increased in many U.S. areas, and zoning reforms expanded into some cities but not others. Studies show that supply constraints result in demand, which increases housing prices. This suggests that when housing markets tighten, homelessness rises due to the high cost of shelter. Building on previous research, our project will compare U.S. cities with different zoning restrictiveness levels and evaluate whether homelessness per capita is wider or narrower when compared with housing prices, which are a cause of zoning restrictions. By combining longitudinal data with cross-city comparisons, our analysis will include prior work as context, along with detailed data from 2015 to 2025.

### Correlational Analysis: Zoning Restrictiveness vs. Homelessness Rate
To directly examine the relationship between our variables, we produced a scatterplot plotting each state's average zoning restrictiveness score against its average homelessness rate per 100,000 residents, averaged across 2015–2024. Points are sized by state population and colored by Census region to reveal whether geographic patterns drive the relationship. A linear regression line is overlaid to assess the direction and strength of the association, with the correlation coefficient and p-value reported to evaluate statistical significance. This plot serves as an initial test of the project's core hypothesis: that states with more restrictive housing policies experience higher per capita homelessness rates

```python
df_avg = df.groupby('state', as_index=False).agg(
    homeless_per_100k=('homeless_per_100k', 'mean'),
    avg_total_restrictions=('avg_total_restrictions', 'mean'),
    pop=('pop', 'mean'),
    state_name=('state_name', 'first'),
    REGION=('REGION', 'first')
)

region_labels = {1: 'Northeast', 2: 'Midwest', 3: 'South', 4: 'West'}
df_avg['region_name'] = df_avg['REGION'].map(region_labels)

fig, ax = plt.subplots(figsize=(12, 8))

palette = {'Northeast': '#e41a1c', 'Midwest': '#377eb8', 'South': '#4daf4a', 'West': '#ff7f00'}
for region, group in df_avg.groupby('region_name'):
    sizes = (group['pop'] / group['pop'].max()) * 800 + 50
    ax.scatter(
        group['avg_total_restrictions'],
        group['homeless_per_100k'],
        s=sizes,
        alpha=0.75,
        label=region,
        color=palette[region],
        edgecolors='white',
        linewidth=0.5
    )
    for _, row in group.iterrows():
        ax.annotate(
            row['state'],
            (row['avg_total_restrictions'], row['homeless_per_100k']),
            fontsize=7, ha='center', va='bottom', alpha=0.8
        )

# ── Regression line ───────────────────────────────────────────────────────────
slope, intercept, r, p, se = stats.linregress(df_avg['avg_total_restrictions'], df_avg['homeless_per_100k'])
x_line = np.linspace(df_avg['avg_total_restrictions'].min(), df_avg['avg_total_restrictions'].max(), 100)
ax.plot(x_line, intercept + slope * x_line, color='black', linewidth=2, linestyle='--', label=f'Regression (r={r:.2f}, p={p:.3f})')

ax.set_xlabel('Average Zoning Restrictiveness Score', fontsize=12)
ax.set_ylabel('Average Homeless per 100k', fontsize=12)
ax.set_title('Zoning Restrictiveness vs. Homelessness Rate by State (2015–2024)\n''(point size = population)', fontsize=13)
ax.legend(title='Region', bbox_to_anchor=(1.01, 1), loc='upper left')
ax.grid(True, alpha=0.3)
plt.tight_layout()
os.makedirs('results', exist_ok=True)
plt.savefig('results/zoning_vs_homelessness_scatter.png', dpi=150, bbox_inches='tight')
plt.show()
print(f"Regression: slope={slope:.2f}, r={r:.2f}, p-value={p:.4f}")
```

## Setup for Modeling

We load the merged state-year panel (490 obs: 49 states × 10 years, DC dropped as a structural outlier with homelessness rates 4× the next-highest state). Six features are engineered before modeling: the outcome (`homeless_per_100k`) is log-transformed to reduce right skew from CA/HI; `year` is mean-centred to prevent multicollinearity with the interaction term; a `zoning × year_c` interaction term is constructed to test whether the zoning effect strengthens over time; `net_mig` is normalised by population to a per-1,000 rate; a `covid` dummy flags the 2020–2021 pandemic shock visible in the EDA time-series; and Census regions are one-hot encoded as fixed effects to absorb the West Coast geographic clustering identified in the choropleth maps.

```python
# install packages
%pip install linearmodels -q
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import statsmodels.api as sm
import statsmodels.formula.api as smf
from statsmodels.stats.diagnostic import het_breuschpagan
from linearmodels.panel import RandomEffects, PooledOLS
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import Ridge
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import cross_val_score, KFold
import warnings
warnings.filterwarnings('ignore')

# ── Load data ─────────────────────────────────────────────────────────────────
df = pd.read_csv('data/02-processed/merged_final.csv')
df = df[df['state'] != 'DC'].copy()   # drop DC — extreme structural outlier

# ── Outcome: log-transform to reduce right skew ───────────────────────────────
df['log_homeless_per_100k'] = np.log(df['homeless_per_100k'])

# ── Centred year: prevents multicollinearity in interaction model ─────────────
df['year_c'] = df['year'] - df['year'].mean()

# ── Interaction term: zoning × centred year ───────────────────────────────────
df['zoning_x_year'] = df['avg_total_restrictions'] * df['year_c']

# ── Net migration rate (per 1,000 residents) ─────────────────────────────────
df['net_mig_rate'] = df['net_mig'] / df['pop'] * 1_000

# ── COVID shock dummy ─────────────────────────────────────────────────────────
df['covid'] = df['year'].isin([2020, 2021]).astype(int)

# ── Region fixed effect dummies ───────────────────────────────────────────────
region_labels = {1: 'Northeast', 2: 'Midwest', 3: 'South', 4: 'West'}
df['region_name'] = df['REGION'].map(region_labels)
df = pd.get_dummies(df, columns=['region_name'], drop_first=True)
region_fe_cols = [c for c in df.columns if c.startswith('region_name_')]

print(f"Dataset: {df.shape[0]} obs | {df['state'].nunique()} states | years {sorted(df['year'].unique())}")
print("\nOutcome variable summary:")
print(df[['homeless_per_100k', 'log_homeless_per_100k']].describe().round(3))
```

## Step 1: OLS Baseline (Bivariate)
We begin with a bivariate OLS baseline regressing log homelessness on zoning restrictiveness alone, with cluster-robust standard errors grouped by state to account for serial correlation across years within each state. This replicates the EDA scatterplot (r = 0.15, p = 0.297) as a formal regression and sets a benchmark R² to beat once controls are added.

```python
X1 = sm.add_constant(df['avg_total_restrictions'])
model1 = sm.OLS(df['log_homeless_per_100k'], X1).fit(
    cov_type='cluster', cov_kwds={'groups': df['state']}
)

print(model1.summary2())

# Convert coefficient to interpretable % effect
b_zoning = model1.params['avg_total_restrictions']
pct_change = (np.exp(b_zoning) - 1) * 100
print(f"\nInterpretation: 1-unit increase in zoning → {pct_change:.1f}% change in homeless/100k")
print(f"R²: {model1.rsquared:.4f}  |  p-value: {model1.pvalues['avg_total_restrictions']:.4f}")
```

### Takeaway
The bivariate baseline explains 11.8% of variance in log homelessness (R² = 0.118). The zoning coefficient is positive and statistically significant (β = 0.579, p = 0.0005), meaning a one-unit increase in average zoning restrictiveness is associated with a 78.4% increase in homelessness per 100k. However, this result should be interpreted cautiously — with only one predictor, the coefficient is almost certainly inflated by omitted variable bias. States with high zoning scores also tend to have high housing prices and sit on the West Coast, so the zoning coefficient is absorbing the combined effect of all three. The large skew and significant Jarque-Bera statistic confirm the residuals are non-normal, reinforcing the need for the log transformation and motivating the addition of controls in Step 2.

## Step 2: Pooled OLS with Full Controls and Region Fixed Effects

The main inferential model adds `avg_hpi`, `net_mig_rate`, a `covid` dummy, a linear year trend, and region fixed effects as controls, isolating the zoning coefficient from the confounders identified in the EDA. Cluster-robust standard errors are again grouped by state. A Breusch-Pagan test is run to confirm whether heteroskedasticity is present and whether the robust SEs are necessary.

```python
formula_controls = (
    "log_homeless_per_100k ~ "
    "avg_total_restrictions + "
    "avg_hpi + "
    "net_mig_rate + "
    "covid + "
    "year_c + "
    + " + ".join(region_fe_cols)
)

model2 = smf.ols(formula_controls, data=df).fit(
    cov_type='cluster', cov_kwds={'groups': df['state']}
)
print(model2.summary2())

# ── Breusch-Pagan test for heteroskedasticity ─────────────────────────────────
bp_stat, bp_pval, _, _ = het_breuschpagan(model2.resid, model2.model.exog)
print(f"\nBreusch-Pagan: stat={bp_stat:.3f}, p={bp_pval:.4f}")
print("→ p < 0.05 confirms heteroskedasticity; cluster-robust SEs are appropriate"
      if bp_pval < 0.05 else "→ Homoskedastic; standard SEs would also be valid")

# ── Readable coefficient table ────────────────────────────────────────────────
key_vars = ['avg_total_restrictions', 'avg_hpi', 'net_mig_rate', 'covid', 'year_c']
coef_df = pd.DataFrame({
    'Coef'  : model2.params[key_vars].round(4),
    'SE'    : model2.bse[key_vars].round(4),
    'p-val' : model2.pvalues[key_vars].round(4),
    '% effect': ((np.exp(model2.params[key_vars]) - 1) * 100).round(2),
    'Sig'   : ['***' if p < 0.01 else '**' if p < 0.05 else
               '*' if p < 0.10 else ''
               for p in model2.pvalues[key_vars]]
})
print("\nKey Coefficients:")
print(coef_df.to_string())
```

### Takeaway
Adding controls more than quadruples the explained variance (R² = 0.508), confirming that the bivariate result was heavily confounded. The picture changes substantially for zoning: once HPI and region are held constant, the zoning coefficient collapses to near zero and becomes statistically insignificant (β = −0.088, p = 0.696). This is the key finding of the model — zoning does not have a direct independent effect on homelessness after controlling for housing prices. The most likely explanation is mediation: zoning raises prices, and prices raise homelessness, but zoning itself has no additional effect once prices are accounted for.

The significant predictors tell a clearer story. Housing prices are the primary driver — a one-point increase in HPI is associated with a 0.08% increase in homelessness per 100k (β = 0.0007, p = 0.023), which compounds meaningfully across the large HPI range in the dataset. Being in the West adds a 78% homelessness premium over the South baseline (β = 0.581, p = 0.001), capturing the structural West Coast effect identified in the choropleth maps. Notably, the covid coefficient is negative and highly significant (β = −0.176, p < 0.001), meaning 2020–2021 homelessness counts were actually 16% *lower* than the trend predicted — likely reflecting eviction moratoriums and emergency shelter expansions temporarily suppressing measured homelessness during those years.

The Breusch-Pagan test confirms heteroskedasticity (p < 0.001), validating the use of cluster-robust standard errors. The high condition number (2,389) flags multicollinearity, most likely between `avg_hpi` and the region dummies given that Western states dominate the high-HPI range — this does not bias the coefficients but widens standard errors, making it harder to detect a zoning effect even if one exists.

## Step 3: Random Effects Panel Model

We use a Random Effects (RE) model rather than Fixed Effects because `avg_total_restrictions` is time-invariant — it does not change year-to-year within states, so state fixed effects would absorb it entirely. RE instead uses both within- and between-state variation to estimate the zoning coefficient while partially controlling for unobserved state heterogeneity. The dataset is restructured to a `(state, year)` MultiIndex as required by `linearmodels`, and standard errors are clustered by state.

```python
panel_df = df.set_index(['state', 'year'])

re_vars = ['avg_total_restrictions', 'avg_hpi', 'net_mig_rate',
           'covid', 'year_c'] + region_fe_cols

Y_panel = panel_df['log_homeless_per_100k']
X_panel = sm.add_constant(panel_df[re_vars])

# Fit Random Effects model with entity-clustered standard errors
re_model   = RandomEffects(Y_panel, X_panel)
re_results = re_model.fit(cov_type='clustered', cluster_entity=True)
print(re_results.summary)

# ── Compare pooled R² vs RE R² (overall) ─────────────────────────────────────
pooled = PooledOLS(Y_panel, X_panel)
pooled_res = pooled.fit(cov_type='clustered', cluster_entity=True)

print(f"\nPooled OLS  R² (benchmark) : {pooled_res.rsquared:.4f}")
print(f"Random Effects R² (overall): {re_results.rsquared:.4f}")
print(f"Random Effects R² (between): {re_results.rsquared_between:.4f}")
print(f"Random Effects R² (within) : {re_results.rsquared_within:.4f}")

# ── Variance decomposition: how much of total variance is between-state? ──────
theta = re_results.theta.mean().item()
print(f"\nTheta (RE weight, 0=pooled, 1=FE): {theta:.4f}")
print("→ Values near 0 mean between-state variance dominates (good for RE)")
```

### Takeaway
The RE model tells a consistent story with Step 2. Zoning remains insignificant (β = −0.123, p = 0.602), HPI is the dominant driver (β = 0.0009, p < 0.001), and the West region premium holds (β = 0.540, p = 0.001). The covid dummy is again negative and significant, confirming the pandemic suppression effect.

The R² decomposition is the most informative output here. The between-state R² of 0.537 means the model explains 54% of the variation in average homelessness *across* states — which is the relevant metric for the research question. The within-state R² of 0.249 captures how well the model tracks year-to-year changes *within* individual states, which is lower since zoning doesn't vary over time and contributes nothing to within-state dynamics.

Theta = 0.853 indicates the RE estimator is weighting heavily toward the Fixed Effects end of the spectrum, meaning there is substantial unobserved state heterogeneity in the data. This is expected given how different states are structurally, and reinforces that the between-state zoning variation — not within-state changes — is doing all the explanatory work. The consistency of the zoning coefficient across Steps 2 and 3 (both near zero, both insignificant) strengthens confidence that the null result for zoning is genuine rather than a modelling artefact.

## Step 4: Interaction Model (Zoning x Year)
We extend the Random Effects (RE) model by adding an interaction between avg_total_restrictions and a centered year variable (year_c) to directly test whether the relationship between zoning restrictions and homelessness changes over time. A positive interaction coefficient indicates that the association between zoning restrictions and homelessness becomes stronger over time, while a negative coefficient suggests the relationship weakens. Centering the year variable reduces multicollinearity and allows the main zoning coefficient to be interpreted at the average year in the sample. The data remain structured as a (state, year) MultiIndex and standard errors are clustered by state.

```python
interaction_vars = ['avg_total_restrictions','year_c','zoning_x_year','avg_hpi','net_mig_rate','covid'] + region_fe_cols

Y_panel = panel_df['log_homeless_per_100k']
X_panel = sm.add_constant(panel_df[interaction_vars])

interaction_model = RandomEffects(Y_panel, X_panel)
interaction_results = interaction_model.fit(cov_type='clustered', cluster_entity=True)

print(interaction_results.summary)

# ── Marginal Effect of Zoning Over Time ──────────────────────────────────────
b_zoning = interaction_results.params['avg_total_restrictions']
b_interaction = interaction_results.params['zoning_x_year']

years = [2015, 2018, 2021, 2024]

print("\nMarginal Effect of Zoning Restrictions")

for yr in years:
    yr_c = yr - df['year'].mean()
    effect = b_zoning + b_interaction * yr_c
    print(f"Year {yr}: {effect:.4f}")
```

### Takeaway 

The interaction between zoning restrictions and year is positive but not statistically significant (β = 0.0072, p = 0.644). While the estimated effect of zoning restrictions becomes slightly less negative over time, there is insufficient evidence to conclude that the relationship between zoning restrictions and homelessness changed meaningfully during the study period. Overall, the results suggest that the association between zoning restrictions and homelessness remained relatively stable from 2015 to 2024.

## Step 5: Random Forest + Ridge CV Nonlinear Benchmark

The previous models focus on estimating and interpreting the relationship between zoning restrictions and homelessness under a linear specification. As a robustness check, we compare their predictive performance against two machine learning models: Ridge Regression and Random Forest Regression. Ridge Regression allows for coefficient shrinkage and improved out-of-sample prediction while retaining a linear structure, whereas Random Forest can capture nonlinear relationships and complex interactions without requiring them to be specified a priori. Five-fold cross-validation is used to evaluate predictive performance. Comparing these models to the Random Effects specification provides insight into how much predictive power may be left unexplained by linear assumptions, while Random Forest feature importance highlights which variables contribute most to prediction.

```python
# ── Use same covariates as RE model (excluding interaction term) ─────────────
ml_vars = ['avg_total_restrictions','avg_hpi','net_mig_rate','covid','year_c'] + region_fe_cols

X = df[ml_vars]
y = df['log_homeless_per_100k']

# ── Ridge Regression: regularized linear benchmark ───────────────────────────
ridge_pipeline = Pipeline([('scaler', StandardScaler()),('ridge', Ridge(alpha=1.0))])

cv = KFold(n_splits=5, shuffle=True, random_state=42)

# ── Five-fold cross-validation ───────────────────────────────────────────────
ridge_scores = cross_val_score(ridge_pipeline, X, y, cv=cv, scoring='r2')

# ── Random Forest: nonlinear benchmark ───────────────────────────────────────
rf = RandomForestRegressor(n_estimators=500, min_samples_leaf=3, random_state=42)

rf_scores = cross_val_score(rf, X, y, cv=cv, scoring='r2')

# ── Compare predictive performance results across the models ──────────────────
print(f"\nRandom Effects R²     : {re_results.rsquared:.4f}")
print(f"Ridge CV Mean R²      : {ridge_scores.mean():.4f}")
print(f"Random Forest CV Mean : {rf_scores.mean():.4f}")

rf.fit(X, y)

importance_df = (pd.DataFrame({'feature': X.columns,'importance': rf.feature_importances_}).sort_values('importance', ascending=False))

print(importance_df)
```

### Takeaway 

The Random Forest model substantially outperforms both the Random Effects and Ridge models (R² ≈ 0.72 vs. 0.29 and 0.47), indicating that homelessness is driven by important nonlinear relationships and interactions that linear specifications miss. The strongest predictors are housing prices (HPI), housing restrictions, and net migration, followed by persistent regional differences—especially in the West. Overall, the results suggest that structural housing market conditions and geography are the primary drivers, rather than time shocks like COVID or simple linear trends.
